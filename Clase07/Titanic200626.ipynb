```python
# Importar las librerias necesarias
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar mis datos
data = pd.read_csv("2titanic.csv")
```

```python
col_names = data.columns.tolist() #Convierta los nombres de las columnas en una lista y las guarde en la variable col_names
col_names
```

```python
for column in col_names:
    print("Valores nulos en", column, ":", data[column].isnull().sum()) #Esta funcion hace una sumatoria de todos los valores nulos de cada columna
```

```python
# Vamos a revisar la estadística descriptiva de mi dataset
data.describe()
```

```python
data.head()
```

```python
def sex_transform(x):
    d = {'male':'M','female':'F'} # Esta funcion va a utilizar este diccionario que al recibir un valor que va a ser la clave y va a retornar el valor correspondient
    return(d[x])

data['Sex'] = data['Sex'].apply(sex_transform)
```

```python
data.head()
```

```python
# Copiar la columna sex a una columna nueva que se llame Gender

data['Gender'] = data['Sex']
data.head()
```

```python
# Cuántos y cuáles pasajeros se subieron sin pagar?

data[data.Fare == 0]
```

```python
# Tabla cruzada entre supervivencia y sexo
pd.crosstab(data.Survived, data.Sex)
```

```python
# Cuantos sobrevivieron en total
data[data.Survived==1].count()
```

```python
# Agrupar los datos por clase y sexo, después solo mostrar la suma de los que sobrevivieron
pclass_sex_survival_count = data.groupby(['Pclass','Sex'])['Survived'].sum()
pclass_sex_survival_count
```

```python
# Tabla cruzada entre Pclass y Sexo
pd.crosstab(data.Pclass, data.Sex)
```

```python
# Gráfica de la columna survided
data.Survived.value_counts().plot(kind = 'bar')
plt.title("Sobrevivieron / cuenta total")
```

```python
# Gráfica de la columna survived por porcentaje
data.Survived.value_counts(normalize = True).plot(kind = "bar", alpha = 0.4)
plt.title("Sobrevivieron / porcentaje total")
```

```python
# Gráfica de sexo y survived en cuenta total
data.Sex[data.Survived == 1].value_counts().plot(kind = 'barh', color = "pink", alpha = 0.7)
plt.title("sobrevivieron totales - M vs F")
```

```python
# Gráfica de sexo y survived en porcentajes
data.Sex[data.Survived == 1].value_counts(normalize = True).plot(kind = 'barh', color = "pink", alpha = 0.9)
plt.title("sobrevivieron porcentajes - M vs F")
```

```python
# Gráfica de supervivencia por clase
data.Pclass[data.Survived == 1].value_counts().plot(kind = "bar", color = "purple")
plt.title("Sobrevivientes por clase - Totales")
```

```python
# Gráfica de supervivencia por clase en porcentajes
data.Pclass[data.Survived == 1].value_counts(normalize = True).plot(kind = "bar", color = "purple", alpha = 0.7)
plt.title("Sobrevivientes por clase - Porcentajes")
```

```python
# Gráfica de supervivencia por clase en porcentajes - pastel
data.Pclass[data.Survived == 1].value_counts(normalize = True).plot(kind = "pie")
plt.title("Sobrevivientes por clase - Porcentajes")
```

```python
# Gráfica para ver si hay una relación entre tipo de ticket y edad
fig = plt.figure(figsize = (20,10))
for clase in [1,2,3]:
    data.Age[data.Pclass == clase].plot(kind = "kde")
plt.legend(("1ra Clase","2da Clase","3era Clase"))
```
