```python
# Librerias

import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn import model_selection
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
```

```python
# Cargar datos
data = pd.read_csv('usuarios_win_mac_lin.csv')
```

```python
# Ver los primeros renglones con head
data.head()
```

```python
# Revisar forma
data.shape
```

```python
# Estadistica descriptiva
data.describe()
```

```python
# Ver cuantos renglones tengo de cada clase (sistemas operativos)
data.groupby('clase').size()
```

```python
# Preparación de mis datos, matriz de entrenamiento "X" que contiene las variables independientes y un vector de valores (labels) "y" que contiene la variable dependiente
X = np.array(data.drop(['clase'], axis = 1))
```

```python
y = np.array(data['clase'])
```

```python
X.shape
```

```python
# Crear modelo predictivo
model = linear_model.LogisticRegression()
```

```python
# Entrenar modelo
model.fit(X,y)
```

```python
# Crear predicciones
predictions = model.predict(X)
```

```python
predictions
```

```python
y
```

```python
model.score(X,y)
```

```python
# BUENAS PRACTICAS
# Separar el dataset en un conjunto de entrenamiento y un conjunto de validación o prueba
validation_size = 0.20
seed = 7
X_train, X_validation, y_train, y_validation = model_selection.train_test_split(X,y, test_size = validation_size, random_state = seed)
```

```python
# Entrenar de nuevo el modelo
name = 'Logistic Regression'
kfold = model_selection.KFold(n_splits=10)
cv_results = model_selection.cross_val_score(model, X_train, y_train, cv=kfold, scoring= 'accuracy')
msg = name, ": ", cv_results.mean(), "(", cv_results.std()
print(msg)
```

```python
# Predicción del conjunto de validación
predictions = model.predict(X_validation)
```

```python
# Evaluar el modelo con los datos que no vio en el entrenamiento
accuracy_score(y_validation, predictions)
```

```python
# Matriz de confusión 
confusion_matrix(y_validation, predictions)
```

```python
print(classification_report(y_validation, predictions))
```
